```python
#Sort 0s, 1s and 2s
class Solution:
    def sort012(self, arr):
        low = 0
        mid = 0
        high = len(arr) - 1

        while mid <= high:
            if arr[mid] == 0:
                arr[low], arr[mid] = arr[mid], arr[low]
                low += 1
                mid += 1
            elif arr[mid] == 1:
                mid += 1
            else:
                arr[mid], arr[high] = arr[high], arr[mid]
                high -= 1

# Input
arr = [0, 1, 2, 0, 1, 2]

# Create object and sort
sol = Solution()
sol.sort012(arr)

# Output
print(arr)
```

```python
#First Occurrence in Sorted
class Solution:
    def binarysearch(self, arr, k):
        left = 0
        right = len(arr) - 1
        result = -1

        while left <= right:
            mid = (left + right) // 2

            if arr[mid] == k:
                result = mid
                right = mid - 1
            elif arr[mid] < k:
                left = mid + 1
            else:
                right = mid - 1

        return result

# Test data
arr = [1, 2, 3, 4, 5]
k = 4

# Create object and call function
sol = Solution()
print(sol.binarysearch(arr, k))

```
